# 📈 Hands-on 2: Regresión Lineal Simple (SLR)
### Análisis Exploratorio y Evaluación de Métricas


## 1. Fundamentos de la Técnica SLR

La **Regresión Lineal Simple** asume que existe una relación de proporcionalidad directa (lineal) entre una variable predictora ($X$) y una variable de respuesta ($Y$).

Antes de aplicar este modelo, es crucial **verificar visualmente** si los datos realmente tienen un comportamiento lineal. Si los datos forman una curva en forma de "U", un modelo de regresión lineal simple no será de mucha ayuda, y necesitaríamos técnicas más complejas (como la regresión polinomial que vimos en sesiones anteriores).

## 2. Modelo Matemático

El modelo se define con la ecuación matemática de una línea recta:

$$ y = m x + c $$

* **$y$**: Variable dependiente (lo que predecimos).
* **$x$**: Variable independiente (nuestro dato de entrada).
* **$m$ (Pendiente / Slope)**: Cuánto aumenta o disminuye $y$ por cada unidad de $x$.
* **$c$ (Intersección / Intercept)**: El valor base de $y$ cuando $x$ es cero.

El objetivo del algoritmo OLS (*Ordinary Least Squares*) es encontrar los valores perfectos de $m$ y $c$ que minimicen la suma de los errores al cuadrado.

## 3. Pipeline de Machine Learning

### Paso 1: Importar librerías
Explicación del código que veremos a continuación:
* `pandas as pd`: Nos permitirá manejar nuestros datos en formato de tabla (DataFrame).
* `matplotlib.pyplot as plt` y `seaborn as sns`: Librerías para crear los gráficos que nos ayudarán en la exploración de los datos.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("✅ Librerías base cargadas.")
```

### Paso 2: Recolección y Análisis Exploratorio de Datos (EDA)
Explicación del código que veremos a continuación:
* `np.random.seed(42)` y generación de datos: Creamos un dataset sintético de Salario vs Experiencia (replicando el usado en el artículo).
* `df.describe()`: Función de pandas que nos da un resumen estadístico (media, valores mínimos y máximos).
* `sns.histplot(...)`: Crea un histograma para ver cómo se distribuyen los salarios (si hay muchos sueldos bajos o altos).
* `sns.scatterplot(...)`: Dibuja un gráfico de puntos para verificar si la relación entre años y salario parece una línea recta.

```python
import numpy as np

# 1. Generamos un dataset sintético para el tutorial
np.random.seed(42)
experiencia = np.linspace(1, 15, 50)
salario = 25000 + (experiencia * 8500) + (np.random.randn(50) * 6000)
df = pd.DataFrame({'Experiencia': experiencia, 'Salario': salario})

# 2. Resumen estadístico
print("--- Resumen Estadístico ---")
display(df.describe())

# 3. Análisis visual: Distribución de la variable objetivo (Salario)
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1) # Crear un panel de 1 fila y 2 columnas, gráfico 1
sns.histplot(df['Salario'], kde=True, color='skyblue')
plt.title('Distribución de Salarios')

# 4. Análisis visual: Relación lineal entre X e Y
plt.subplot(1, 2, 2) # Gráfico 2
sns.scatterplot(data=df, x='Experiencia', y='Salario', color='orange')
plt.title('Experiencia vs Salario')

plt.tight_layout()
plt.show()
```

### Paso 3: Separar variables y dividir en Train / Test
Explicación del código que veremos a continuación:
* `X = df[['Experiencia']].values`: Extraemos la columna 'Experiencia' como una matriz bidimensional (requisito de Scikit-Learn).
* `y = df['Salario'].values`: Extraemos el 'Salario' como nuestro vector a predecir.
* `train_test_split(...)`: Parte los datos. `test_size=0.2` indica que 20% de los datos se guardan para evaluar el modelo al final.

```python
from sklearn.model_selection import train_test_split

# Variables independientes (X) y dependientes (y)
X = df[['Experiencia']].values
y = df['Salario'].values

# División de los datos
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

print(f"Datos divididos: {X_train.shape[0]} para entrenar, {X_test.shape[0]} para probar.")
```

### Paso 4 y 5: Entrenamiento y Predicción
Explicación del código que veremos a continuación:
* `LinearRegression()`: Inicializa el modelo matemático de regresión lineal.
* `modelo.fit(X_train, y_train)`: Aplica el algoritmo a los datos de entrenamiento para encontrar la pendiente y la intersección.
* `modelo.predict(X_test)`: Usa la ecuación matemática recién descubierta para intentar adivinar los salarios de los datos que apartamos (el 20% de prueba).

```python
from sklearn.linear_model import LinearRegression

# Inicialización y entrenamiento
modelo = LinearRegression()
modelo.fit(X_train, y_train)

# Generación de predicciones
y_pred = modelo.predict(X_test)

print("✅ Modelo entrenado y predicciones generadas.")
print(f"Ecuación: Salario = {modelo.intercept_:.2f} + {modelo.coef_[0]:.2f} * Experiencia")
```

### Paso 6: Visualización Final de la Recta de Regresión
Explicación del código que veremos a continuación:
* `plt.scatter(X_train, y_train)`: Pinta los puntos reales que el modelo usó para aprender.
* `plt.plot(X_train, modelo.predict(X_train))`: Pinta la línea recta matemática que el modelo dedujo. Si la línea atraviesa el centro de la nube de puntos, el ajuste es bueno.

```python
plt.figure(figsize=(8,5))

# Puntos de entrenamiento en gris
plt.scatter(X_train, y_train, color='gray', label='Datos Entrenamiento')
# Puntos de prueba en rojo
plt.scatter(X_test, y_test, color='red', label='Datos Prueba (Nuevos)')
# La línea de regresión en azul
plt.plot(X_train, modelo.predict(X_train), color='blue', linewidth=2, label='Línea de Ajuste')

plt.title('Regresión Lineal: Salario vs Experiencia')
plt.xlabel('Años de Experiencia')
plt.ylabel('Salario Calculado')
plt.legend()
plt.show()
```

### Paso 7: Evaluación del Modelo (Métricas)
*(Diferenciador principal respecto al Tutorial 1)*
La inspección visual no es suficiente en Machine Learning. Necesitamos números duros para saber qué tan bueno es el modelo.
Explicación del código que veremos a continuación:
* `mean_squared_error(y_test, y_pred)`: Calcula el Error Cuadrático Medio. Es el promedio de los errores al cuadrado. Queremos que sea lo más bajo posible.
* `r2_score(y_test, y_pred)`: Calcula el coeficiente de determinación (R²). Va de 0 a 1. Indica qué porcentaje de la variación de los salarios es explicada por la experiencia. Si es 0.90, significa que el modelo es muy bueno (90% de precisión explicativa).

```python
from sklearn.metrics import mean_squared_error, r2_score

# Cálculos matemáticos del error
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("--- RESULTADOS DE EVALUACIÓN (Métricas) ---")
print(f"Error Cuadrático Medio (MSE): {mse:,.2f}")
print(f"Coeficiente R²: {r2:.4f} (El modelo explica el {r2*100:.2f}% de los datos)")
```
